# TorchCode vs TorchLeet

All **41 TorchCode exercises**, including **1–11**, compared with the local TorchLeet question descriptions and templates. Rechecked **November 1, 2026**.

- **Direct / same implementation:** TorchLeet asks you to implement the same operation, either separately or inside a larger exercise. Interfaces, variants and tests can differ.
- **Related:** a different variant or task teaches some of the same ideas.
- Use only: the model uses the operation, usually through a built-in layer; that does not replace implementing its formula from scratch.
- **None found:** no dedicated implementation exercise found in the local question files.

**Linear attention (12)** still has no dedicated local TorchLeet match found. Flash Attention, sliding-window attention and Mamba are distinct techniques.

## What the overlap means for studying

For **1–11**, the clearest implementation overlaps are **2, 5, 6, 8, 9, 10 and 11**. TorchLeet uses or relates to **1, 3, 4 and 7**, but does not provide the same standalone from-scratch exercises for those.

Across **1–41**, no dedicated implementation match was found for **1, 3, 4, 7, 12, 16, 17, 18, 19, 21, 30 or 31**. Some are used extensively in larger TorchLeet models. **15 (SwiGLU), 36 (manual INT8 linear) and 41 (OPD)** also cover specific implementations that their related TorchLeet exercises do not replace.

A matching topic does not mean identical difficulty or requirements. In particular, using nn.LayerNorm/nn.GELU/nn.Dropout is different from writing their formulas, and implementing one component of a capstone is different from completing the whole capstone.
